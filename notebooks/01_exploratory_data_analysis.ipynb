# Exploratory Data Analysis (PayPulse)

This notebook loads the synthetic transaction dataset and provides basic statistics and visualisations to understand the data distribution and relationships relevant for the recovery model.


```python
import pandas as pdimport matplotlib.pyplot as pltimport seaborn as sns%matplotlib inline
```

```python
raw_path = '../data/raw/synthetic_transactions.csv'df = pd.read_csv(raw_path)df.head()
```

## 1. Dataset Overview
- Number of records: `len(df)`
- Columns and dtypes: `df.dtypes`
- Check for missing values: `df.isnull().sum()`

```python
print('Rows:', df.shape[0])print('Columns:', df.shape[1])print(df.dtypes)print('
Missing values per column:')print(df.isnull().sum())
```

## 2. Recovery Rate Distribution
Plot the proportion of successful recoveries vs failures.

```python
recov_rate = df['recovered'].mean()sns.barplot(x=['Failed','Recovered'], y=[1-recov_rate, recov_rate])plt.title('Overall Recovery Rate')plt.ylabel('Proportion')plt.show()
```

## 3. Failure Reason Impact
Compare recovery rates across different failure reasons.

```python
reason_stats = df.groupby('failure_reason')['recovered'].mean().reset_index()plt.figure(figsize=(8,4))sns.barplot(data=reason_stats, x='failure_reason', y='recovered')plt.xticks(rotation=45)plt.title('Recovery Rate by Failure Reason')plt.ylabel('Recovery Probability')plt.show()
```

## 4. Transaction Amount vs Recovery
Investigate whether larger payments have lower recovery probability.

```python
sns.boxplot(x='recovered', y='transaction_amount', data=df)plt.title('Transaction Amount by Recovery Outcome')plt.xlabel('Recovered (0=No, 1=Yes)')plt.show()
```

## 5. Hourly Failure Pattern
Does the hour of day affect recovery?

```python
hour_stats = df.groupby('transaction_hour')['recovered'].mean().reset_index()plt.figure(figsize=(10,4))sns.lineplot(data=hour_stats, x='transaction_hour', y='recovered')plt.title('Recovery Probability by Transaction Hour')plt.xlabel('Hour of Day')plt.ylabel('Recovery Probability')plt.show()
```
